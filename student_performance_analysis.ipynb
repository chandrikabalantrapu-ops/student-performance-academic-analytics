```python
import pandas as pd

# Dataset URL
url = "https://raw.githubusercontent.com/chandrikabalantrapu-ops/student-performance-academic-analytics/refs/heads/main/data/student-mat.csv"

# Load the dataset
df = pd.read_csv(url, sep=";")

# Display first five records
df.head()
```

```python
# 1. Check number of rows and columns
print("Dataset Shape:", df.shape)

# 2. Display all column names
print("\nColumn Names:")
print(df.columns.tolist())

# 3. Check dataset information
print("\nDataset Information:")
df.info()

# 4. Check missing values
print("\nMissing Values:")
print(df.isnull().sum())
```

```python
# STEP 1: Check duplicate records

print("Number of duplicate rows:", df.duplicated().sum())


# STEP 2: Statistical summary of numerical columns

print("\nNumerical Data Summary:")
display(df.describe())


# STEP 3: Check unique values in important columns

print("\nUnique values in important columns:")

columns_to_check = ['age', 'studytime', 'failures', 'absences', 'G1', 'G2', 'G3']

for column in columns_to_check:
    print(f"\n{column}:")
    print(sorted(df[column].unique()))
```

```python
# Create student performance categories

def categorize_performance(marks):

    if marks < 10:
        return "Needs Support"

    elif marks < 15:
        return "Average"

    else:
        return "Good"


df['performance_category'] = df['G3'].apply(
    categorize_performance
)

# Count students in each category

print(df['performance_category'].value_counts())
```

```python
import statsmodels.formula.api as smf
```

```python
model = smf.ols(
    formula="""
    G3 ~ G1 + G2
    + C(schoolsup)
    + C(famsup)
    + C(paid)
    + C(internet)
    """,
    data=df
).fit()

print(model.summary())
```

```python
# Calculate predicted grades and residuals

df['predicted_G3'] = model.predict(df)

df['residuals'] = df['G3'] - df['predicted_G3']

print("Actual grades:")
print(df['G3'].head())

print("\nPredicted grades:")
print(df['predicted_G3'].head())

print("\nResiduals:")
print(df['residuals'].head())
```

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    x=df['predicted_G3'],
    y=df['residuals']
)

plt.axhline(y=0, color='red', linestyle='--')

plt.title("Residual Plot")
plt.xlabel("Predicted Final Grades")
plt.ylabel("Residuals")

plt.show()
```

```python
import statsmodels.api as sm

sm.qqplot(model.resid, line='45', fit=True)

plt.title("Q-Q Plot of Regression Residuals")
plt.show()
```

```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error
from sklearn.metrics import mean_squared_error, r2_score
```

```python
# Select input features
features = [
    'G1',
    'G2',
    'schoolsup',
    'famsup',
    'paid',
    'internet'
]

X = df[features]

# Target variable
y = df['G3']

# Convert categorical columns into numerical columns
X = pd.get_dummies(
    X,
    drop_first=True,
    dtype=int
)

print(X.head())
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("Training records:", X_train.shape[0])
print("Testing records:", X_test.shape[0])
```

```python
# Create model
lr_model = LinearRegression()

# Train model
lr_model.fit(X_train, y_train)

# Predict test data
y_pred = lr_model.predict(X_test)

print("Model training completed!")
```

```python
mae = mean_absolute_error(y_test, y_pred)

rmse = np.sqrt(mean_squared_error(y_test, y_pred))

r2 = r2_score(y_test, y_pred)

print("Model Evaluation Results")

print("MAE:", round(mae, 3))

print("RMSE:", round(rmse, 3))

print("R2 Score:", round(r2, 3))
```

```python
comparison = pd.DataFrame({
    'Actual Grades': y_test.values,
    'Predicted Grades': y_pred
})

print(comparison.head(15))
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error
from sklearn.metrics import mean_squared_error, r2_score
import numpy as np

# Model A: Only previous examination grades

model_a = LinearRegression()

model_a.fit(
    X_train[['G1', 'G2']],
    y_train
)

pred_a = model_a.predict(
    X_test[['G1', 'G2']]
)

print("MODEL A: G1 and G2")

print("MAE:", round(
    mean_absolute_error(y_test, pred_a), 3
))

print("RMSE:", round(
    np.sqrt(mean_squared_error(y_test, pred_a)), 3
))

print("R2:", round(
    r2_score(y_test, pred_a), 3
))
```

```python
# Model B: All six encoded predictors

pred_b = lr_model.predict(X_test)

results = pd.DataFrame({
    'Model': ['Model A: G1 and G2',
              'Model B: All Features'],

    'MAE': [
        mean_absolute_error(y_test, pred_a),
        mean_absolute_error(y_test, pred_b)
    ],

    'RMSE': [
        np.sqrt(mean_squared_error(y_test, pred_a)),
        np.sqrt(mean_squared_error(y_test, pred_b))
    ],

    'R2 Score': [
        r2_score(y_test, pred_a),
        r2_score(y_test, pred_b)
    ]
})

print(results.round(3))
```

```python
from sklearn.model_selection import cross_val_score

cv_scores = cross_val_score(
    lr_model,
    X_train,
    y_train,
    cv=5,
    scoring='r2'
)

print("Cross-validation R2 scores:")
print(cv_scores)

print("Average CV R2:",
      round(cv_scores.mean(), 3))

print("Standard deviation:",
      round(cv_scores.std(), 3))
```
